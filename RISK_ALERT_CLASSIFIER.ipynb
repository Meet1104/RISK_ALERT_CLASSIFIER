# Part A — Conceptual Understanding

## Question 1: What is Logistic Regression and why is it suitable for classification?

- Logistic Regression is a supervised machine learning algorithm used to predict the probability of an outcome belonging to a particular class.
- It uses the logistic (sigmoid) function to convert predictions into probabilities between 0 and 1.
- It is suitable for classification problems, especially binary classification, such as predicting whether a customer is low-risk or high-risk.


## Question 2: Classification Metrics & Why Accuracy Alone Is Insufficient

- Classification metrics are used to evaluate how well a classification model predicts the correct classes.
- Common metrics include Accuracy, Precision, Recall, F1-Score, and Confusion Matrix.
- Accuracy alone can be misleading when the dataset is imbalanced, because a model may correctly predict the majority class but fail to identify minority-class samples.
- Therefore, multiple evaluation metrics should be used to assess model performance effectively.

## Question 3: Type-I Error and Type-II Error in Risk Prediction

- Type-I Error (False Positive): A Type-I Error occurs when the model incorrectly predicts a low-risk customer as high-risk. In risk prediction, this may result in unnecessary alerts, additional verification, or rejection of a customer who is actually financially reliable.

- Type-II Error (False Negative): A Type-II Error occurs when the model incorrectly predicts a high-risk customer as low-risk. This may result in financial losses for the bank because a potentially defaulting customer is not identified.

## Question 4: Precision, Recall, F1-Score, TPR & FPR

- Precision measures the proportion of customers predicted as high-risk who are actually high-risk.

- Recall measures the proportion of actual high-risk customers correctly identified by the model.

- F1-Score is the harmonic mean of Precision and Recall. It balances both metrics and is useful when the dataset is imbalanced.

- True Positive Rate (TPR) measures the proportion of actual positive cases correctly identified. It is equivalent to Recall.

- False Positive Rate (FPR) measures the proportion of actual negative cases incorrectly classified as positive.


## Question 5: What is AUC-ROC and how does it help in evaluating

- ROC (Receiver Operating Characteristic) Curve is a graphical representation of a classification model's performance at different classification thresholds. It plots the True Positive Rate (TPR) against the False Positive Rate (FPR).

- AUC (Area Under the Curve) measures the area under the ROC curve and represents the model's ability to distinguish between positive and negative classes. A higher AUC indicates better class discrimination. AUC-ROC is useful for evaluating classifiers across different thresholds and comparing model performance.

## Question 6: Why Does Class Imbalance Create Problems in Classification

- Class imbalance occurs when the number of observations in different classes of a dataset is not evenly distributed. For example, in a risk prediction dataset, there may be significantly more low-risk customers than high-risk customers.

- Class imbalance can create problems because a machine learning model may become biased toward the majority class and perform poorly when predicting the minority class. As a result, the model may achieve high accuracy while failing to identify important minority-class observations.

- To address class imbalance, techniques such as under-sampling, over-sampling, SMOTE, and ADASYN can be used. Evaluation metrics such as Recall, Precision, F1-Score, and ROC-AUC are also important for evaluating model performance.

- The **Risk Alert Classifier** is a supervised machine learning project designed to predict whether a customer is **Low Risk (0)** or **High Risk (1)** using demographic, financial, behavioral, and transaction-related data.

- The dataset contains **4,600 customer records**, with 4,043 Low-Risk customers and 557 High-Risk customers, making it an imbalanced classification dataset.

- The project involved data understanding, missing-value analysis, preprocessing, train-test splitting, and building a Logistic Regression baseline model. To handle class imbalance, Under-Sampling, Over-Sampling, SMOTE, and ADASYN were applied and compared. Decision Tree and Random Forest classifiers were also implemented and evaluated.

- Randomized Search CV and Grid Search CV were used for hyperparameter tuning. Models were compared using Accuracy, Precision, Recall, F1-Score, Confusion Matrix, and AUC-ROC, along with ROC curve analysis.

- The primary objective was to select a model with strong High-Risk customer detection, **high Recall, and fewer False Negatives**, while maintaining reasonable Precision and overall performance.

- The project demonstrates how supervised machine learning can help identify potentially risky customers and support better risk-management decisions.
